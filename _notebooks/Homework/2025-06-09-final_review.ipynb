## Project work

### Bathroom Pass (OCS)
Developed a digital check-in and check-out hall pass system for Poway Unified School District that reduces student roaming during class time by 30%, engages teachers and students in real-time tracking, and enhances classroom attendance through features like automated alerts for unauthorized exits, a user-friendly interface for students to request passes, and data analytics to identify peak roaming times, thereby increasing overall class time engagement and fostering a more focused learning environment.

#### Personal contributions
- Made the queuing system for students wanting to leave the classroom
- Created SpringBoot backend for viewing data (admin only)
- Worked on UI changes --> made it data viewing and using the bathroom pass easy to use

### Titanic ML (AP CSA)
I developed a machine learning project focused on predicting passenger survival on the Titanic using Java-based tools. The goal was to apply data science and machine learning principles within the Java ecosystem, leveraging Weka for model building and Tablesaw for data preprocessing and visualization.

#### Personal contributions
- Performed data wrangling and exploratory analysis with Tablesaw, including handling missing values, feature encoding, and basic statistics.
- Implemented and evaluated multiple ML models using Weka, including J48 decision trees.

### Educeri Webscraping (personal)
I developed an automated web scraper to extract structured data from an educational website, with full permission from the site owner. This project was aimed to collect lessons for students in my non-profit organization to use

#### Learnings/Takeaways
- Used Selenium WebDriver to programmatically navigate the website and interact with web elements like buttons, dropdowns, and forms.
- Extracted relevant educational content including course titles, descriptions, and sorted by grade level

### Charged Fluid Simulator (personal)
A Python-based simulation of electrostatic plasma dynamics using a fluid approximation. This project solves Poisson’s equation for electric potential in a 2D grid and evolves the charge density over time using the continuity equation.

#### Learnings/Takeaways
- Currently self-learning multivariable calculus and differential equations
- Making a 2D mesh with Numpy's meshgrid
- Using scipy's functions to make my life easier
